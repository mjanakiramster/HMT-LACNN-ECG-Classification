```python
# =========================================================
# PTB-XL HMT-LACNN — FULL EVALUATION PIPELINE
# Sections:
#   0. Setup & Data Loading (shared)
#   1. Proper Test Set Evaluation
#   2. Baseline Comparisons
#   3. Ablation Study
#   4. Statistical Significance (Multi-seed)
#   5. Final Summary
# =========================================================

# ── SECTION 0: SETUP ──────────────────────────────────────
from google.colab import drive
drive.mount('/content/drive')

import os, ast, random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras.layers import (
    Input, Conv1D, Conv2D, BatchNormalization, ReLU,
    Add, GlobalAveragePooling1D, GlobalAveragePooling2D,
    Dense, Dropout, LSTM, Bidirectional, MaxPooling1D
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.metrics import (
    roc_auc_score, f1_score, precision_score,
    recall_score, confusion_matrix
)
from sklearn.preprocessing import MultiLabelBinarizer

# ── PATHS ─────────────────────────────────────────────────
DATA_PATH    = "/content/drive/My Drive/PTBXL_processed/"
META_PATH    = "/content/drive/My Drive/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.3/"
SAVE_ROOT    = "/content/drive/My Drive/ECG-classification/Trained models/"
RESULTS_PATH = "/content/drive/My Drive/ECG-classification/Results/"
BASELINE_PATH= "/content/drive/My Drive/ECG-classification/Baselines/"
ABLATION_PATH= "/content/drive/My Drive/ECG-classification/Ablation/"
MULTISEED_PATH="/content/drive/My Drive/ECG-classification/MultiSeed/"

for p in [RESULTS_PATH, BASELINE_PATH, ABLATION_PATH, MULTISEED_PATH]:
    os.makedirs(p, exist_ok=True)

# ── LOAD RAW ECG ──────────────────────────────────────────
X_train_raw = np.load(DATA_PATH + "X_train.npy", allow_pickle=True)
X_test_raw  = np.load(DATA_PATH + "X_test.npy",  allow_pickle=True)

# 2D shape for HMT-LACNN:  (N, 12, 1000, 1)
X_train_2d = X_train_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)
X_test_2d  = X_test_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)

# 1D shape for baselines:  (N, 1000, 12)  — time steps first, leads second
X_train_1d = X_train_raw  # already (N, 1000, 12) before any transpose
X_test_1d  = X_test_raw

# Safety check — if raw shape is (N, 12, 1000), transpose it
if X_train_1d.shape[1] == 12 and X_train_1d.shape[2] == 1000:
    X_train_1d = X_train_1d.transpose(0, 2, 1)
    X_test_1d  = X_test_1d.transpose(0, 2, 1)

print("2D Train:", X_train_2d.shape, " | 2D Test:", X_test_2d.shape)
print("1D Train:", X_train_1d.shape, " | 1D Test:", X_test_1d.shape)

# ── LOAD METADATA & BUILD LABELS ──────────────────────────
Y = pd.read_csv(META_PATH + "ptbxl_database.csv", index_col="ecg_id")
Y.scp_codes = Y.scp_codes.apply(lambda x: ast.literal_eval(x))

agg_df = pd.read_csv(META_PATH + "scp_statements.csv", index_col=0)
agg_df = agg_df[agg_df.diagnostic == 1]

def aggregate_superclass(y):
    return list(set(
        agg_df.loc[k].diagnostic_class
        for k in y if k in agg_df.index
    ))

def aggregate_subclass(y):
    return list(set(
        agg_df.loc[k].diagnostic_subclass
        for k in y if k in agg_df.index
    ))

Y["superclass"] = Y.scp_codes.apply(aggregate_superclass)
Y["subclass"]   = Y.scp_codes.apply(aggregate_subclass)

train_idx = Y.strat_fold != 10   # official PTB-XL train
test_idx  = Y.strat_fold == 10   # official PTB-XL test

# Fit binarizers on TRAIN only — transform test separately
mlb_super = MultiLabelBinarizer()
y_super_train = mlb_super.fit_transform(Y[train_idx].superclass.tolist())
y_super_test  = mlb_super.transform(Y[test_idx].superclass.tolist())

mlb_sub = MultiLabelBinarizer()
y_sub_train = mlb_sub.fit_transform(Y[train_idx].subclass.tolist())
y_sub_test  = mlb_sub.transform(Y[test_idx].subclass.tolist())

def to_binary(labels):
    return np.array([1 if 'NORM' in l else 0 for l in labels])

y_bin_train = to_binary(Y[train_idx].superclass.tolist()).reshape(-1, 1)
y_bin_test  = to_binary(Y[test_idx].superclass.tolist())

NUM_SUPER = y_super_train.shape[1]
NUM_SUB   = y_sub_train.shape[1]

print(f"\nSuperclass classes ({NUM_SUPER}): {mlb_super.classes_}")
print(f"Subclass classes  ({NUM_SUB}): {mlb_sub.classes_}")
print(f"Binary  train: {y_bin_train.shape} | test: {y_bin_test.shape}")
print(f"Super   train: {y_super_train.shape} | test: {y_super_test.shape}")
print(f"Sub     train: {y_sub_train.shape} | test: {y_sub_test.shape}")

# ── SHARED UTILITIES ──────────────────────────────────────
def get_callbacks(patience_es=15, patience_lr=5):
    return [
        EarlyStopping(patience=patience_es, restore_best_weights=True),
        ReduceLROnPlateau(patience=patience_lr, factor=0.5, verbose=0)
    ]

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

def compute_metrics(y_true, y_pred_prob, class_names, threshold=0.5):
    y_pred = (y_pred_prob >= threshold).astype(int)
    results = []
    for i, cls in enumerate(class_names):
        yt = y_true[:, i] if y_true.ndim > 1 else y_true
        yp = y_pred[:, i] if y_pred.ndim > 1 else y_pred
        yp_prob = y_pred_prob[:, i] if y_pred_prob.ndim > 1 else y_pred_prob
        try:
            auc = roc_auc_score(yt, yp_prob)
        except Exception:
            auc = float('nan')
        tn, fp, fn, tp = confusion_matrix(yt, yp, labels=[0,1]).ravel()
        results.append({
            "Class":       cls,
            "AUC":         round(auc, 4),
            "Sensitivity": round(tp/(tp+fn) if (tp+fn)>0 else 0, 4),
            "Specificity": round(tn/(tn+fp) if (tn+fp)>0 else 0, 4),
            "F1":          round(f1_score(yt, yp, zero_division=0), 4),
            "Precision":   round(precision_score(yt, yp, zero_division=0), 4),
            "Recall":      round(recall_score(yt, yp, zero_division=0), 4),
            "Support":     int(yt.sum())
        })
    return pd.DataFrame(results)

def bootstrap_auc(y_true, y_pred_prob, n=1000, ci=95, seed=42):
    np.random.seed(seed)
    aucs = []
    N = len(y_true)
    for _ in range(n):
        idx = np.random.choice(N, N, replace=True)
        yt, yp = y_true[idx], y_pred_prob[idx]
        try:
            aucs.append(roc_auc_score(yt, yp,
                         average='macro' if yt.ndim > 1 else None))
        except Exception:
            pass
    lo = np.percentile(aucs, (100-ci)/2)
    hi = np.percentile(aucs, 100-(100-ci)/2)
    return np.mean(aucs), lo, hi

def plot_confusion_matrices(y_true, y_pred_prob, class_names, title, save_path, threshold=0.5):
    y_pred = (y_pred_prob >= threshold).astype(int)
    if y_true.ndim == 1:
        cm = confusion_matrix(y_true, y_pred)
        fig, ax = plt.subplots(figsize=(5, 4))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=['Abnormal','NORM'],
                    yticklabels=['Abnormal','NORM'], ax=ax)
        ax.set_title(title); ax.set_ylabel('True'); ax.set_xlabel('Predicted')
    else:
        n = len(class_names)
        fig, axes = plt.subplots(1, n, figsize=(4*n, 4))
        if n == 1: axes = [axes]
        for i, (cls, ax) in enumerate(zip(class_names, axes)):
            cm = confusion_matrix(y_true[:, i], y_pred[:, i])
            sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                        xticklabels=['Neg','Pos'],
                        yticklabels=['Neg','Pos'], ax=ax)
            ax.set_title(cls, fontsize=9)
        fig.suptitle(title, fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f"  Saved → {save_path}")

# ── HMT-LACNN ARCHITECTURE ────────────────────────────────
def residual_block_2d(x, filters, k):
    shortcut = x
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x)
    if shortcut.shape[-1] != filters:
        shortcut = Conv2D(filters, (1,1))(shortcut)
    x = Add()([x, shortcut])
    return ReLU()(x)

def build_hmt_lacnn(num_super=NUM_SUPER, num_sub=NUM_SUB):
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_2d(x, 64, 5)
    x = residual_block_2d(x, 128, 5)
    x = Conv2D(256, (12,1))(x)                  # Cross-lead fusion
    x = BatchNormalization()(x); x = ReLU()(x)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x)
    x = Dropout(0.4)(x)
    out_bin   = Dense(1,        activation="sigmoid", name="binary")(x)
    out_super = Dense(num_super,activation="sigmoid", name="super")(x)
    out_sub   = Dense(num_sub,  activation="sigmoid", name="sub")(x)
    return Model(inp, [out_bin, out_super, out_sub])

def compile_hmt(model):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(3e-4),
        loss={"binary":"binary_crossentropy",
              "super": "binary_crossentropy",
              "sub":   "binary_crossentropy"},
        metrics={"binary": tf.keras.metrics.AUC(name="auc"),
                 "super":  tf.keras.metrics.AUC(name="auc", multi_label=True),
                 "sub":    tf.keras.metrics.AUC(name="auc", multi_label=True)}
    )
    return model

# ==========================================================
# SECTION 1: PROPER TEST SET EVALUATION
# ==========================================================
print("\n" + "="*60)
print("SECTION 1: PROPER TEST SET EVALUATION")
print("="*60)

# Load your pre-trained model
main_model = tf.keras.models.load_model(
    SAVE_ROOT + "Twenty-three-classification/12_leads_2.h5"
)

pred_bin, pred_super, pred_sub = main_model.predict(
    X_test_2d, batch_size=32, verbose=1
)
pred_bin = pred_bin.squeeze()

# ── Binary head ───────────────────────────────────────────
print("\n--- Binary Head (NORM vs Abnormal) ---")
bin_df = compute_metrics(
    y_bin_test.reshape(-1,1), pred_bin.reshape(-1,1), ["NORM vs Abnormal"]
)
print(bin_df.to_string(index=False))
m, lo, hi = bootstrap_auc(y_bin_test, pred_bin)
print(f"Bootstrap AUC: {m:.4f} (95% CI: {lo:.4f} – {hi:.4f})")

# ── Superclass head ───────────────────────────────────────
print("\n--- Superclass Head (5-class) ---")
super_df = compute_metrics(y_super_test, pred_super, mlb_super.classes_.tolist())
print(super_df.to_string(index=False))
m, lo, hi = bootstrap_auc(y_super_test, pred_super)
print(f"Macro AUC: {roc_auc_score(y_super_test, pred_super, average='macro'):.4f}")
print(f"Bootstrap AUC: {m:.4f} (95% CI: {lo:.4f} – {hi:.4f})")

# ── Subclass head ─────────────────────────────────────────
print("\n--- Subclass Head (23-class) ---")
sub_df = compute_metrics(y_sub_test, pred_sub, mlb_sub.classes_.tolist())
print(sub_df.to_string(index=False))
m, lo, hi = bootstrap_auc(y_sub_test, pred_sub)
print(f"Macro AUC: {roc_auc_score(y_sub_test, pred_sub, average='macro'):.4f}")
print(f"Bootstrap AUC: {m:.4f} (95% CI: {lo:.4f} – {hi:.4f})")

# ── Save CSVs ─────────────────────────────────────────────
bin_df.to_csv(RESULTS_PATH + "metrics_binary.csv", index=False)
super_df.to_csv(RESULTS_PATH + "metrics_superclass.csv", index=False)
sub_df.to_csv(RESULTS_PATH + "metrics_subclass.csv", index=False)

# ── Confusion matrices ────────────────────────────────────
plot_confusion_matrices(y_bin_test, pred_bin, ["Binary"],
    "Confusion Matrix — Binary Head",
    RESULTS_PATH + "cm_binary.png")

plot_confusion_matrices(y_super_test, pred_super, mlb_super.classes_.tolist(),
    "Confusion Matrix — Superclass Head",
    RESULTS_PATH + "cm_superclass.png")

plot_confusion_matrices(y_sub_test, pred_sub, mlb_sub.classes_.tolist(),
    "Confusion Matrix — Subclass Head",
    RESULTS_PATH + "cm_subclass.png")

print("\n✅ Section 1 Complete")

# ==========================================================
# SECTION 2: BASELINE COMPARISONS
# ==========================================================
print("\n" + "="*60)
print("SECTION 2: BASELINE COMPARISONS")
print("="*60)

# ── Baseline architectures ────────────────────────────────
def residual_block_1d(x, filters, k):
    shortcut = x
    x = Conv1D(filters, k, padding='same')(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = Conv1D(filters, k, padding='same')(x)
    x = BatchNormalization()(x)
    if shortcut.shape[-1] != filters:
        shortcut = Conv1D(filters, 1)(shortcut)
    x = Add()([x, shortcut])
    return ReLU()(x)

def build_simple_cnn():
    inp = Input(shape=(1000, 12))
    x = Conv1D(32, 7, padding='same', activation='relu')(inp)
    x = BatchNormalization()(x); x = MaxPooling1D(2)(x)
    x = Conv1D(64, 5, padding='same', activation='relu')(x)
    x = BatchNormalization()(x); x = MaxPooling1D(2)(x)
    x = Conv1D(128, 5, padding='same', activation='relu')(x)
    x = BatchNormalization()(x); x = MaxPooling1D(2)(x)
    x = Conv1D(256, 3, padding='same', activation='relu')(x)
    x = BatchNormalization()(x)
    x = GlobalAveragePooling1D()(x)
    x = Dense(128, activation='relu')(x); x = Dropout(0.4)(x)
    return Model(inp, Dense(NUM_SUPER, activation='sigmoid')(x), name="SimpleCNN")

def build_resnet_1d():
    inp = Input(shape=(1000, 12))
    x = Conv1D(64, 15, padding='same')(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_1d(x, 64, 7)
    x = residual_block_1d(x, 128, 5)
    x = residual_block_1d(x, 256, 5)
    x = GlobalAveragePooling1D()(x)
    x = Dense(128, activation='relu')(x); x = Dropout(0.4)(x)
    return Model(inp, Dense(NUM_SUPER, activation='sigmoid')(x), name="ResNet1D")

def build_bilstm():
    inp = Input(shape=(1000, 12))
    x = Conv1D(64, 7, padding='same', activation='relu')(inp)
    x = BatchNormalization()(x); x = MaxPooling1D(4)(x)
    x = Bidirectional(LSTM(64, return_sequences=True))(x)
    x = Bidirectional(LSTM(32))(x)
    x = Dense(64, activation='relu')(x); x = Dropout(0.4)(x)
    return Model(inp, Dense(NUM_SUPER, activation='sigmoid')(x), name="BiLSTM")

baseline_results = []
baselines = [
    ("SimpleCNN", build_simple_cnn),
    ("ResNet1D",  build_resnet_1d),
    ("BiLSTM",    build_bilstm),
]

for name, build_fn in baselines:
    print(f"\n  Training: {name}")
    tf.keras.backend.clear_session()
    m = build_fn()
    m.compile(
        optimizer=tf.keras.optimizers.Adam(3e-4),
        loss='binary_crossentropy',
        metrics=[tf.keras.metrics.AUC(name='auc', multi_label=True)]
    )
    m.fit(X_train_1d, y_super_train,
          validation_split=0.1, epochs=100, batch_size=32,
          callbacks=get_callbacks(patience_es=10), verbose=1)
    pred = m.predict(X_test_1d, batch_size=32, verbose=0)
    auc  = roc_auc_score(y_super_test, pred, average='macro')
    print(f"  {name} Test Macro-AUC: {auc:.4f}")
    baseline_results.append({"Model": name, "Macro_AUC": round(auc, 4)})
    m.save(BASELINE_PATH + f"{name}.h5")

# Add our model
our_auc = roc_auc_score(y_super_test, pred_super, average='macro')
baseline_results.append({"Model": "HMT-LACNN (Ours)", "Macro_AUC": round(our_auc, 4)})

df_baselines = pd.DataFrame(baseline_results).sort_values("Macro_AUC", ascending=False)
print("\n--- Baseline Comparison Table ---")
print(df_baselines.to_string(index=False))
df_baselines.to_csv(RESULTS_PATH + "baseline_comparison.csv", index=False)

# Bar chart
colors = ['#E91E63' if 'Ours' in m else '#90CAF9' for m in df_baselines["Model"]]
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(df_baselines["Model"], df_baselines["Macro_AUC"],
              color=colors, edgecolor='black')
ax.set_ylim(df_baselines["Macro_AUC"].min() - 0.03, 1.0)
ax.set_ylabel("Macro AUC (Test Set)", fontsize=12)
ax.set_title("Model Comparison — PTB-XL Superclass", fontsize=13, fontweight='bold')
for bar, val in zip(bars, df_baselines["Macro_AUC"]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.001,
            f'{val:.4f}', ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.savefig(RESULTS_PATH + "baseline_comparison.png", dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ Section 2 Complete")

# ==========================================================
# SECTION 3: ABLATION STUDY
# ==========================================================
print("\n" + "="*60)
print("SECTION 3: ABLATION STUDY")
print("="*60)

def plain_block_2d(x, filters, k):
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x)
    return ReLU()(x)

def build_variant_B():
    """No multi-task — single head only"""
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_2d(x, 64, 5)
    x = residual_block_2d(x, 128, 5)
    x = Conv2D(256, (12,1))(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x); x = Dropout(0.4)(x)
    out = Dense(NUM_SUPER, activation="sigmoid", name="super")(x)
    return Model(inp, out, name="B_NoMultiTask")

def build_variant_C():
    """No cross-lead fusion — remove (12,1) conv"""
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_2d(x, 64, 5)
    x = residual_block_2d(x, 128, 5)
    # Skip cross-lead fusion
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x); x = Dropout(0.4)(x)
    out_bin   = Dense(1,         activation="sigmoid", name="binary")(x)
    out_super = Dense(NUM_SUPER, activation="sigmoid", name="super")(x)
    out_sub   = Dense(NUM_SUB,   activation="sigmoid", name="sub")(x)
    return Model(inp, [out_bin, out_super, out_sub], name="C_NoCrossLead")

def build_variant_D():
    """No residual blocks — plain CNN"""
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = plain_block_2d(x, 64, 5)
    x = plain_block_2d(x, 128, 5)
    x = Conv2D(256, (12,1))(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x); x = Dropout(0.4)(x)
    out_bin   = Dense(1,         activation="sigmoid", name="binary")(x)
    out_super = Dense(NUM_SUPER, activation="sigmoid", name="super")(x)
    out_sub   = Dense(NUM_SUB,   activation="sigmoid", name="sub")(x)
    return Model(inp, [out_bin, out_super, out_sub], name="D_NoResidual")

ablation_results = [
    {"Variant": "A_Full (Ours)", "Description": "Full model",
     "Macro_AUC": round(our_auc, 4)}
]

ablation_variants = [
    ("B_NoMultiTask",  "No multi-task (single head)", build_variant_B, "single"),
    ("C_NoCrossLead",  "No cross-lead fusion",        build_variant_C, "multi"),
    ("D_NoResidual",   "No residual blocks",          build_variant_D, "multi"),
]

for name, desc, build_fn, mode in ablation_variants:
    print(f"\n  Ablation: {name} — {desc}")
    tf.keras.backend.clear_session()
    m = build_fn()

    if mode == "multi":
        m = compile_hmt(m)
        m.fit(X_train_2d,
              {"binary": y_bin_train, "super": y_super_train, "sub": y_sub_train},
              validation_split=0.1, epochs=100, batch_size=16,
              callbacks=get_callbacks(), verbose=1)
        preds = m.predict(X_test_2d, batch_size=32, verbose=0)
        pred_s = preds[1]
    else:
        m.compile(optimizer=tf.keras.optimizers.Adam(3e-4),
                  loss='binary_crossentropy',
                  metrics=[tf.keras.metrics.AUC(name='auc', multi_label=True)])
        m.fit(X_train_2d, y_super_train,
              validation_split=0.1, epochs=100, batch_size=16,
              callbacks=get_callbacks(), verbose=1)
        pred_s = m.predict(X_test_2d, batch_size=32, verbose=0)

    auc = roc_auc_score(y_super_test, pred_s, average='macro')
    print(f"  {name} Macro-AUC: {auc:.4f}")
    ablation_results.append({"Variant": name, "Description": desc,
                              "Macro_AUC": round(auc, 4)})
    m.save(ABLATION_PATH + f"{name}.h5")

df_ablation = pd.DataFrame(ablation_results)
full_auc = df_ablation.loc[df_ablation.Variant=="A_Full (Ours)", "Macro_AUC"].values[0]
df_ablation["Delta"] = (df_ablation["Macro_AUC"] - full_auc).round(4)

print("\n--- Ablation Study Results ---")
print(df_ablation.to_string(index=False))
df_ablation.to_csv(RESULTS_PATH + "ablation_results.csv", index=False)

colors = ['#E91E63' if 'Ours' in v else '#90CAF9' for v in df_ablation["Variant"]]
fig, ax = plt.subplots(figsize=(9, 5))
bars = ax.bar(df_ablation["Variant"], df_ablation["Macro_AUC"],
              color=colors, edgecolor='black')
ax.axhline(full_auc, color='red', linestyle='--', linewidth=1.5, label='Full model')
ax.set_ylim(df_ablation["Macro_AUC"].min() - 0.02, 1.0)
ax.set_ylabel("Macro AUC (Test Set)", fontsize=12)
ax.set_title("Ablation Study — Component Contribution", fontsize=13, fontweight='bold')
for bar, val in zip(bars, df_ablation["Macro_AUC"]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.001,
            f'{val:.4f}', ha='center', va='bottom', fontsize=10)
ax.legend(); plt.tight_layout()
plt.savefig(RESULTS_PATH + "ablation_study.png", dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ Section 3 Complete")

# ==========================================================
# SECTION 4: STATISTICAL SIGNIFICANCE (5 SEEDS)
# ==========================================================
print("\n" + "="*60)
print("SECTION 4: STATISTICAL SIGNIFICANCE (5 SEEDS)")
print("="*60)

SEEDS = [42, 123, 256, 789, 1024]
seed_results = []

for seed in SEEDS:
    print(f"\n  Seed: {seed}")
    set_seed(seed)
    tf.keras.backend.clear_session()

    m = build_hmt_lacnn()
    m = compile_hmt(m)
    m.fit(X_train_2d,
          {"binary": y_bin_train, "super": y_super_train, "sub": y_sub_train},
          validation_split=0.1, epochs=100, batch_size=16,
          callbacks=get_callbacks(), verbose=1)

    p_bin, p_sup, p_sub = m.predict(X_test_2d, batch_size=32, verbose=0)
    p_bin = p_bin.squeeze()

    auc_b = roc_auc_score(y_bin_test,   p_bin, )
    auc_s = roc_auc_score(y_super_test, p_sup, average='macro')
    auc_u = roc_auc_score(y_sub_test,   p_sub, average='macro')

    print(f"  Binary: {auc_b:.4f} | Super: {auc_s:.4f} | Sub: {auc_u:.4f}")
    seed_results.append({"Seed": seed,
                          "AUC_Binary": round(auc_b, 4),
                          "AUC_Super":  round(auc_s, 4),
                          "AUC_Sub":    round(auc_u, 4)})
    m.save(MULTISEED_PATH + f"model_seed_{seed}.h5")

df_seeds = pd.DataFrame(seed_results)
df_seeds.to_csv(RESULTS_PATH + "multiseed_results.csv", index=False)

print("\n--- Multi-Seed Results ---")
print(df_seeds.to_string(index=False))
print("\n--- Mean ± Std (Publication Format) ---")
for col in ["AUC_Binary", "AUC_Super", "AUC_Sub"]:
    print(f"  {col}: {df_seeds[col].mean():.4f} ± {df_seeds[col].std():.4f}")

# Stability plot
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(SEEDS, df_seeds["AUC_Binary"], 'o-', label='Binary',     color='#E91E63')
ax.plot(SEEDS, df_seeds["AUC_Super"],  's-', label='Superclass', color='#2196F3')
ax.plot(SEEDS, df_seeds["AUC_Sub"],    '^-', label='Subclass',   color='#4CAF50')
for col, color in zip(["AUC_Binary","AUC_Super","AUC_Sub"],
                       ['#E91E63','#2196F3','#4CAF50']):
    ax.axhline(df_seeds[col].mean(), linestyle='--', color=color, alpha=0.4)
ax.set_xlabel("Random Seed", fontsize=12)
ax.set_ylabel("AUC", fontsize=12)
ax.set_title("HMT-LACNN Stability Across 5 Seeds", fontsize=13, fontweight='bold')
ax.set_xticks(SEEDS); ax.legend(); ax.set_ylim(0.85, 1.0)
plt.tight_layout()
plt.savefig(RESULTS_PATH + "multiseed_stability.png", dpi=150, bbox_inches='tight')
plt.show()

print("\n✅ Section 4 Complete")

# ==========================================================
# SECTION 5: FINAL SUMMARY + LATEX TABLES
# ==========================================================
print("\n" + "="*60)
print("SECTION 5: FINAL SUMMARY")
print("="*60)

print("\n[1] Baseline Comparison")
print(df_baselines.to_string(index=False))

print("\n[2] Ablation Study")
print(df_ablation[["Variant","Description","Macro_AUC","Delta"]].to_string(index=False))

print("\n[3] Multi-Seed Summary (mean ± std)")
for col in ["AUC_Binary", "AUC_Super", "AUC_Sub"]:
    print(f"  {col}: {df_seeds[col].mean():.4f} ± {df_seeds[col].std():.4f}")

# ── LaTeX table: multi-seed ───────────────────────────────
print("\n--- LATEX: Multi-Seed Table (paste into paper) ---")
print(r"\begin{table}[h]\centering")
print(r"\caption{HMT-LACNN Performance on PTB-XL Test Set (5 random seeds)}")
print(r"\begin{tabular}{lccc}\hline")
print(r"Seed & Binary AUC & Superclass AUC & Subclass AUC \\\hline")
for _, row in df_seeds.iterrows():
    print(f"{int(row.Seed)} & {row.AUC_Binary:.4f} & {row.AUC_Super:.4f} & {row.AUC_Sub:.4f} \\\\")
mb,sb = df_seeds.AUC_Binary.mean(), df_seeds.AUC_Binary.std()
ms,ss = df_seeds.AUC_Super.mean(),  df_seeds.AUC_Super.std()
mu,su = df_seeds.AUC_Sub.mean(),    df_seeds.AUC_Sub.std()
print(r"\hline")
print(f"Mean±Std & {mb:.4f}±{sb:.4f} & {ms:.4f}±{ss:.4f} & {mu:.4f}±{su:.4f} \\\\")
print(r"\hline\end{tabular}\end{table}")

# ── LaTeX table: baseline comparison ─────────────────────
print("\n--- LATEX: Baseline Comparison Table ---")
print(r"\begin{table}[h]\centering")
print(r"\caption{Comparison with Baseline Models on PTB-XL Superclass Task}")
print(r"\begin{tabular}{lc}\hline")
print(r"Model & Macro AUC \\\hline")
for _, row in df_baselines.iterrows():
    bold = r"\textbf{" + f"{row.Macro_AUC:.4f}" + "}" if "Ours" in row.Model else f"{row.Macro_AUC:.4f}"
    print(f"{row.Model} & {bold} \\\\")
print(r"\hline\end{tabular}\end{table}")

print("\n" + "="*60)
print("✅ ALL SECTIONS COMPLETE")
print(f"   Results saved to: {RESULTS_PATH}")
print("="*60)
```

```python
# =========================================================
# SECTION 4 RECOVERY — Continue from Seed 123
# Seed 42 already completed — resume from 123
# =========================================================
seed_results = [
    {"Seed": 42, "AUC_Binary": 0.9514, "AUC_Super": 0.9223, "AUC_Sub": 0.8836}
]
from google.colab import drive
drive.mount('/content/drive')

import os, ast, random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, ReLU,
    Add, GlobalAveragePooling2D, Dense, Dropout
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import MultiLabelBinarizer

# ── PATHS ─────────────────────────────────────────────────
DATA_PATH     = "/content/drive/My Drive/PTBXL_processed/"
META_PATH     = "/content/drive/My Drive/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.3/"
MULTISEED_PATH= "/content/drive/My Drive/ECG-classification/MultiSeed/"
RESULTS_PATH  = "/content/drive/My Drive/ECG-classification/Results/"
os.makedirs(MULTISEED_PATH, exist_ok=True)
os.makedirs(RESULTS_PATH, exist_ok=True)

# ── LOAD DATA ─────────────────────────────────────────────
X_train_raw = np.load(DATA_PATH + "X_train.npy", allow_pickle=True)
X_test_raw  = np.load(DATA_PATH + "X_test.npy",  allow_pickle=True)

X_train_2d = X_train_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)
X_test_2d  = X_test_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)

print("Train:", X_train_2d.shape, "| Test:", X_test_2d.shape)

# ── LABELS ────────────────────────────────────────────────
Y = pd.read_csv(META_PATH + "ptbxl_database.csv", index_col="ecg_id")
Y.scp_codes = Y.scp_codes.apply(lambda x: ast.literal_eval(x))
agg_df = pd.read_csv(META_PATH + "scp_statements.csv", index_col=0)
agg_df = agg_df[agg_df.diagnostic == 1]

def aggregate_superclass(y):
    return list(set(agg_df.loc[k].diagnostic_class for k in y if k in agg_df.index))
def aggregate_subclass(y):
    return list(set(agg_df.loc[k].diagnostic_subclass for k in y if k in agg_df.index))

Y["superclass"] = Y.scp_codes.apply(aggregate_superclass)
Y["subclass"]   = Y.scp_codes.apply(aggregate_subclass)

train_idx = Y.strat_fold != 10
test_idx  = Y.strat_fold == 10

mlb_super = MultiLabelBinarizer()
y_super_train = mlb_super.fit_transform(Y[train_idx].superclass.tolist())
y_super_test  = mlb_super.transform(Y[test_idx].superclass.tolist())

mlb_sub = MultiLabelBinarizer()
y_sub_train = mlb_sub.fit_transform(Y[train_idx].subclass.tolist())
y_sub_test  = mlb_sub.transform(Y[test_idx].subclass.tolist())

y_bin_train = np.array([1 if 'NORM' in l else 0
                        for l in Y[train_idx].superclass.tolist()]).reshape(-1, 1)
y_bin_test  = np.array([1 if 'NORM' in l else 0
                        for l in Y[test_idx].superclass.tolist()])

NUM_SUPER = y_super_train.shape[1]
NUM_SUB   = y_sub_train.shape[1]
print(f"Superclass: {NUM_SUPER} | Subclass: {NUM_SUB}")

# ── MODEL ─────────────────────────────────────────────────
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

def residual_block_2d(x, filters, k):
    shortcut = x
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x)
    if shortcut.shape[-1] != filters:
        shortcut = Conv2D(filters, (1,1))(shortcut)
    x = Add()([x, shortcut])
    return ReLU()(x)

def build_hmt_lacnn():
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_2d(x, 64, 5)
    x = residual_block_2d(x, 128, 5)
    x = Conv2D(256, (12,1))(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x)
    x = Dropout(0.4)(x)
    out_bin   = Dense(1,         activation="sigmoid", name="binary")(x)
    out_super = Dense(NUM_SUPER, activation="sigmoid", name="super")(x)
    out_sub   = Dense(NUM_SUB,   activation="sigmoid", name="sub")(x)
    return Model(inp, [out_bin, out_super, out_sub])

def compile_hmt(model):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(3e-4),
        loss={"binary":"binary_crossentropy",
              "super": "binary_crossentropy",
              "sub":   "binary_crossentropy"},
        metrics={"binary": tf.keras.metrics.AUC(name="auc"),
                 "super":  tf.keras.metrics.AUC(name="auc", multi_label=True),
                 "sub":    tf.keras.metrics.AUC(name="auc", multi_label=True)}
    )
    return model

def get_callbacks():
    return [
        EarlyStopping(patience=15, restore_best_weights=True),
        ReduceLROnPlateau(patience=5, factor=0.5, verbose=0)
    ]

# ── LOAD SEED 42 RESULT (already completed) ───────────────
# Seed 42 result from your logs
seed_results = [
    {"Seed": 42, "AUC_Binary": 0.9514, "AUC_Super": 0.9223, "AUC_Sub": 0.8836}
]
print("Loaded Seed 42 result from logs: ", seed_results[0])

# ── CONTINUE FROM SEED 123 ────────────────────────────────
REMAINING_SEEDS = [123, 256, 789, 1024]

for seed in REMAINING_SEEDS:
    print(f"\n{'='*60}")
    print(f"  Seed: {seed}")
    print(f"{'='*60}")

    set_seed(seed)
    tf.keras.backend.clear_session()

    m = build_hmt_lacnn()
    m = compile_hmt(m)

    m.fit(
        X_train_2d,
        {"binary": y_bin_train, "super": y_super_train, "sub": y_sub_train},
        validation_split=0.1,
        epochs=100,
        batch_size=16,
        callbacks=get_callbacks(),
        verbose=1
    )

    p_bin, p_sup, p_sub = m.predict(X_test_2d, batch_size=32, verbose=0)
    p_bin = p_bin.squeeze()

    auc_b = roc_auc_score(y_bin_test,   p_bin)
    auc_s = roc_auc_score(y_super_test, p_sup, average='macro')
    auc_u = roc_auc_score(y_sub_test,   p_sub, average='macro')

    print(f"\n  Seed {seed} → Binary: {auc_b:.4f} | Super: {auc_s:.4f} | Sub: {auc_u:.4f}")

    seed_results.append({
        "Seed": seed,
        "AUC_Binary": round(auc_b, 4),
        "AUC_Super":  round(auc_s, 4),
        "AUC_Sub":    round(auc_u, 4)
    })

    m.save(MULTISEED_PATH + f"model_seed_{seed}.h5")

    # ✅ Save after EVERY seed so timeout can't lose results
    df_partial = pd.DataFrame(seed_results)
    df_partial.to_csv(RESULTS_PATH + "multiseed_results.csv", index=False)
    print(f"  ✅ Saved results so far ({len(seed_results)}/5 seeds complete)")

# ── FINAL SUMMARY ─────────────────────────────────────────
df_seeds = pd.DataFrame(seed_results)
print("\n" + "="*60)
print("MULTI-SEED FINAL RESULTS")
print("="*60)
print(df_seeds.to_string(index=False))

print("\n--- Mean ± Std (Publication Format) ---")
for col in ["AUC_Binary", "AUC_Super", "AUC_Sub"]:
    print(f"  {col}: {df_seeds[col].mean():.4f} ± {df_seeds[col].std():.4f}")

# ── STABILITY PLOT ────────────────────────────────────────
SEEDS = df_seeds["Seed"].tolist()
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(SEEDS, df_seeds["AUC_Binary"], 'o-', label='Binary',     color='#E91E63')
ax.plot(SEEDS, df_seeds["AUC_Super"],  's-', label='Superclass', color='#2196F3')
ax.plot(SEEDS, df_seeds["AUC_Sub"],    '^-', label='Subclass',   color='#4CAF50')
for col, color in zip(["AUC_Binary","AUC_Super","AUC_Sub"],
                       ['#E91E63','#2196F3','#4CAF50']):
    ax.axhline(df_seeds[col].mean(), linestyle='--', color=color, alpha=0.4)
ax.set_xlabel("Random Seed", fontsize=12)
ax.set_ylabel("AUC", fontsize=12)
ax.set_title("HMT-LACNN Stability Across 5 Seeds", fontsize=13, fontweight='bold')
ax.set_xticks(SEEDS)
ax.legend()
ax.set_ylim(0.85, 1.0)
plt.tight_layout()
plt.savefig(RESULTS_PATH + "multiseed_stability.png", dpi=150, bbox_inches='tight')
plt.show()

# ── LATEX TABLE ───────────────────────────────────────────
print("\n--- LATEX TABLE (paste into paper) ---")
print(r"\begin{table}[h]\centering")
print(r"\caption{HMT-LACNN Performance on PTB-XL Test Set (5 random seeds)}")
print(r"\begin{tabular}{lccc}\hline")
print(r"Seed & Binary AUC & Superclass AUC & Subclass AUC \\\hline")
for _, row in df_seeds.iterrows():
    print(f"{int(row.Seed)} & {row.AUC_Binary:.4f} & {row.AUC_Super:.4f} & {row.AUC_Sub:.4f} \\\\")
mb,sb = df_seeds.AUC_Binary.mean(), df_seeds.AUC_Binary.std()
ms,ss = df_seeds.AUC_Super.mean(),  df_seeds.AUC_Super.std()
mu,su = df_seeds.AUC_Sub.mean(),    df_seeds.AUC_Sub.std()
print(r"\hline")
print(f"Mean$\pm$Std & {mb:.4f}$\pm${sb:.4f} & {ms:.4f}$\pm${ss:.4f} & {mu:.4f}$\pm${su:.4f} \\\\")
print(r"\hline\end{tabular}\end{table}")

print("\n✅ SECTION 4 COMPLETE — All 5 seeds done")
```

```python
# ==========================================================
# MINIMAL SETUP — Run this FIRST in the new Colab account
# ==========================================================
from google.colab import drive
drive.mount('/content/drive')

import os, ast, random
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from tensorflow.keras.layers import (
    Input, Conv1D, Conv2D, BatchNormalization, ReLU,
    Add, GlobalAveragePooling1D, GlobalAveragePooling2D,
    Dense, Dropout, LSTM, Bidirectional, MaxPooling1D
)
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.metrics import (
    roc_auc_score, f1_score, precision_score,
    recall_score, confusion_matrix, accuracy_score,
    matthews_corrcoef, balanced_accuracy_score
)
from sklearn.preprocessing import MultiLabelBinarizer

# ── PATHS ──────────────────────────────────────────────────
DATA_PATH     = "/content/drive/My Drive/PTBXL_processed/"
META_PATH     = "/content/drive/My Drive/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.3/"
SAVE_ROOT     = "/content/drive/My Drive/ECG-classification/Trained models/"
RESULTS_PATH  = "/content/drive/My Drive/ECG-classification/Results/"
BASELINE_PATH = "/content/drive/My Drive/ECG-classification/Baselines/"
ABLATION_PATH = "/content/drive/My Drive/ECG-classification/Ablation/"
MULTISEED_PATH= "/content/drive/My Drive/ECG-classification/MultiSeed/"

for p in [RESULTS_PATH, BASELINE_PATH, ABLATION_PATH, MULTISEED_PATH]:
    os.makedirs(p, exist_ok=True)

# ── LOAD ECG DATA ──────────────────────────────────────────
X_train_raw = np.load(DATA_PATH + "X_train.npy", allow_pickle=True)
X_test_raw  = np.load(DATA_PATH + "X_test.npy",  allow_pickle=True)

X_train_2d = X_train_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)
X_test_2d  = X_test_raw.transpose(0, 2, 1).reshape(-1, 12, 1000, 1)

print("2D Train:", X_train_2d.shape, "| 2D Test:", X_test_2d.shape)

# ── LOAD METADATA & BUILD LABELS ───────────────────────────
Y = pd.read_csv(META_PATH + "ptbxl_database.csv", index_col="ecg_id")
Y.scp_codes = Y.scp_codes.apply(lambda x: ast.literal_eval(x))
agg_df = pd.read_csv(META_PATH + "scp_statements.csv", index_col=0)
agg_df = agg_df[agg_df.diagnostic == 1]

def aggregate_superclass(y):
    return list(set(agg_df.loc[k].diagnostic_class for k in y if k in agg_df.index))
def aggregate_subclass(y):
    return list(set(agg_df.loc[k].diagnostic_subclass for k in y if k in agg_df.index))

Y["superclass"] = Y.scp_codes.apply(aggregate_superclass)
Y["subclass"]   = Y.scp_codes.apply(aggregate_subclass)

train_idx = Y.strat_fold != 10
test_idx  = Y.strat_fold == 10

mlb_super = MultiLabelBinarizer()
y_super_train = mlb_super.fit_transform(Y[train_idx].superclass.tolist())
y_super_test  = mlb_super.transform(Y[test_idx].superclass.tolist())

mlb_sub = MultiLabelBinarizer()
y_sub_train = mlb_sub.fit_transform(Y[train_idx].subclass.tolist())
y_sub_test  = mlb_sub.transform(Y[test_idx].subclass.tolist())

def to_binary(labels):
    return np.array([1 if 'NORM' in l else 0 for l in labels])

y_bin_train = to_binary(Y[train_idx].superclass.tolist()).reshape(-1, 1)
y_bin_test  = to_binary(Y[test_idx].superclass.tolist())

NUM_SUPER = y_super_train.shape[1]
NUM_SUB   = y_sub_train.shape[1]

print(f"Superclass classes ({NUM_SUPER}): {mlb_super.classes_}")
print(f"Subclass classes  ({NUM_SUB}): {mlb_sub.classes_}")

# ── SHARED UTILITIES ───────────────────────────────────────
def get_callbacks(patience_es=15, patience_lr=5):
    return [
        EarlyStopping(patience=patience_es, restore_best_weights=True),
        ReduceLROnPlateau(patience=patience_lr, factor=0.5, verbose=0)
    ]

def set_seed(seed):
    random.seed(seed); np.random.seed(seed); tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

# ── HMT-LACNN ARCHITECTURE ─────────────────────────────────
def residual_block_2d(x, filters, k):
    shortcut = x
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = Conv2D(filters, (1, k), padding="same")(x)
    x = BatchNormalization()(x)
    if shortcut.shape[-1] != filters:
        shortcut = Conv2D(filters, (1,1))(shortcut)
    x = Add()([x, shortcut])
    return ReLU()(x)

def build_hmt_lacnn(num_super=NUM_SUPER, num_sub=NUM_SUB):
    inp = Input(shape=(12, 1000, 1))
    x = Conv2D(32, (1,7), padding="same")(inp)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = residual_block_2d(x, 64, 5)
    x = residual_block_2d(x, 128, 5)
    x = Conv2D(256, (12,1))(x)
    x = BatchNormalization()(x); x = ReLU()(x)
    x = GlobalAveragePooling2D()(x)
    x = Dense(256, activation="relu")(x); x = Dropout(0.4)(x)
    out_bin   = Dense(1,        activation="sigmoid", name="binary")(x)
    out_super = Dense(num_super,activation="sigmoid", name="super")(x)
    out_sub   = Dense(num_sub,  activation="sigmoid", name="sub")(x)
    return Model(inp, [out_bin, out_super, out_sub])

def compile_hmt(model):
    model.compile(
        optimizer=tf.keras.optimizers.Adam(3e-4),
        loss={"binary":"binary_crossentropy","super":"binary_crossentropy","sub":"binary_crossentropy"},
        metrics={"binary": tf.keras.metrics.AUC(name="auc"),
                 "super":  tf.keras.metrics.AUC(name="auc", multi_label=True),
                 "sub":    tf.keras.metrics.AUC(name="auc", multi_label=True)}
    )
    return model

print("\n✅ Setup complete — now run the Section 4 Resume cell")
```

```python
# ==========================================================
# ROC CURVES — Load model and draw for all 3 heads
# ==========================================================
from sklearn.metrics import roc_curve, auc as sk_auc
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
import warnings
warnings.filterwarnings('ignore')

# ── Load model and get predictions ────────────────────────
print("Loading model and generating predictions...")
main_model = tf.keras.models.load_model(
    SAVE_ROOT + "Twenty-three-classification/12_leads_2.h5"
)
pred_bin, pred_super, pred_sub = main_model.predict(
    X_test_2d, batch_size=32, verbose=1
)
pred_bin = pred_bin.squeeze()
print("✅ Predictions ready")

# ══════════════════════════════════════════════════════════
# FIGURE 1: Main 3-panel ROC (publication figure)
# ══════════════════════════════════════════════════════════
fig = plt.figure(figsize=(21, 7))
gs  = gridspec.GridSpec(1, 3, figure=fig, wspace=0.32)

# ── Panel 1: Binary Head ──────────────────────────────────
ax1 = fig.add_subplot(gs[0])

fpr, tpr, thresholds = roc_curve(y_bin_test, pred_bin)
roc_auc_val = sk_auc(fpr, tpr)
opt_idx = np.argmax(tpr - fpr)
opt_thresh = thresholds[opt_idx]

ax1.plot(fpr, tpr, color='#E91E63', lw=2.5,
         label=f'AUC = {roc_auc_val:.4f}')
ax1.fill_between(fpr, tpr, alpha=0.08, color='#E91E63')
ax1.plot([0,1],[0,1], 'k--', lw=1, alpha=0.5)
ax1.scatter(fpr[opt_idx], tpr[opt_idx],
            color='#E91E63', s=100, zorder=6, marker='*')
ax1.annotate(
    f'Optimal\n(FPR={fpr[opt_idx]:.2f}, TPR={tpr[opt_idx]:.2f})\nThreshold={opt_thresh:.2f}',
    xy=(fpr[opt_idx], tpr[opt_idx]),
    xytext=(fpr[opt_idx]+0.12, tpr[opt_idx]-0.15),
    fontsize=8.5, color='#333333',
    arrowprops=dict(arrowstyle='->', color='gray', lw=1.2),
    bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8)
)
ax1.set_xlim([0,1]); ax1.set_ylim([0,1.02])
ax1.set_xlabel('False Positive Rate', fontsize=12)
ax1.set_ylabel('True Positive Rate', fontsize=12)
ax1.set_title('Binary Head\n(NORM vs Abnormal)', fontsize=13, fontweight='bold')
ax1.legend(loc='lower right', fontsize=11)
ax1.grid(True, alpha=0.3)

# Add metrics box
bin_metrics = f'Sensitivity: 0.9045\nSpecificity: 0.8567\nF1: 0.8662\nMCC: 0.7561'
ax1.text(0.97, 0.35, bin_metrics, transform=ax1.transAxes,
         fontsize=8.5, verticalalignment='top', horizontalalignment='right',
         bbox=dict(boxstyle='round', facecolor='#FFF3F7', alpha=0.9, edgecolor='#E91E63'))

# ── Panel 2: Superclass Head (5 classes) ─────────────────
ax2 = fig.add_subplot(gs[1])

colors_super = ['#E91E63','#2196F3','#4CAF50','#FF9800','#9C27B0']
super_classes = mlb_super.classes_.tolist()
macro_auc_super = roc_auc_score(y_super_test, pred_super, average='macro')

# Per-class curves
for i, (cls, color) in enumerate(zip(super_classes, colors_super)):
    fpr_i, tpr_i, _ = roc_curve(y_super_test[:, i], pred_super[:, i])
    auc_i = sk_auc(fpr_i, tpr_i)
    ax2.plot(fpr_i, tpr_i, color=color, lw=2,
             label=f'{cls} (AUC={auc_i:.3f})')
    ax2.fill_between(fpr_i, tpr_i, alpha=0.03, color=color)

# Macro average curve
all_fpr = np.unique(np.concatenate([
    roc_curve(y_super_test[:,i], pred_super[:,i])[0]
    for i in range(len(super_classes))
]))
mean_tpr = np.zeros_like(all_fpr)
for i in range(len(super_classes)):
    fpr_i, tpr_i, _ = roc_curve(y_super_test[:,i], pred_super[:,i])
    mean_tpr += np.interp(all_fpr, fpr_i, tpr_i)
mean_tpr /= len(super_classes)

ax2.plot(all_fpr, mean_tpr, color='black', lw=2.5, linestyle='--',
         label=f'Macro Avg (AUC={macro_auc_super:.4f})')
ax2.plot([0,1],[0,1], 'k--', lw=1, alpha=0.3)
ax2.set_xlim([0,1]); ax2.set_ylim([0,1.02])
ax2.set_xlabel('False Positive Rate', fontsize=12)
ax2.set_ylabel('True Positive Rate', fontsize=12)
ax2.set_title('Superclass Head\n(5 Diagnostic Classes)', fontsize=13, fontweight='bold')
ax2.legend(loc='lower right', fontsize=9.5)
ax2.grid(True, alpha=0.3)

# ── Panel 3: Subclass Head — Top 10 by AUC ───────────────
ax3 = fig.add_subplot(gs[2])

sub_classes = mlb_sub.classes_.tolist()
macro_auc_sub = roc_auc_score(y_sub_test, pred_sub, average='macro')

# Compute per-class AUC, skip single-class
sub_aucs = []
for i, cls in enumerate(sub_classes):
    if len(np.unique(y_sub_test[:,i])) < 2:
        continue
    fpr_i, tpr_i, _ = roc_curve(y_sub_test[:,i], pred_sub[:,i])
    sub_aucs.append((cls, i, sk_auc(fpr_i, tpr_i)))

sub_aucs_sorted = sorted(sub_aucs, key=lambda x: x[2], reverse=True)
top10 = sub_aucs_sorted[:10]

cmap = plt.cm.tab10
for j, (cls, idx, auc_val) in enumerate(top10):
    fpr_i, tpr_i, _ = roc_curve(y_sub_test[:,idx], pred_sub[:,idx])
    ax3.plot(fpr_i, tpr_i, color=cmap(j/10), lw=1.8,
             label=f'{cls} ({auc_val:.3f})')

# Macro average
all_fpr_sub = np.unique(np.concatenate([
    roc_curve(y_sub_test[:,i], pred_sub[:,i])[0]
    for cls,i,_ in sub_aucs
]))
mean_tpr_sub = np.zeros_like(all_fpr_sub)
for cls, i, _ in sub_aucs:
    fpr_i, tpr_i, _ = roc_curve(y_sub_test[:,i], pred_sub[:,i])
    mean_tpr_sub += np.interp(all_fpr_sub, fpr_i, tpr_i)
mean_tpr_sub /= len(sub_aucs)

ax3.plot(all_fpr_sub, mean_tpr_sub, 'k--', lw=2.5,
         label=f'Macro Avg (AUC={macro_auc_sub:.4f})')
ax3.plot([0,1],[0,1],'k--', lw=1, alpha=0.3)
ax3.set_xlim([0,1]); ax3.set_ylim([0,1.02])
ax3.set_xlabel('False Positive Rate', fontsize=12)
ax3.set_ylabel('True Positive Rate', fontsize=12)
ax3.set_title('Subclass Head\n(Top 10 of 23 Classes by AUC)', fontsize=13, fontweight='bold')
ax3.legend(loc='lower right', fontsize=8.5, ncol=1)
ax3.grid(True, alpha=0.3)

plt.suptitle('HMT-LACNN — ROC Curves on PTB-XL Test Set',
             fontsize=15, fontweight='bold', y=1.01)
plt.savefig(RESULTS_PATH + "roc_curves_main.png", dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Main ROC figure saved")

# ══════════════════════════════════════════════════════════
# FIGURE 2: All 23 subclasses (separate figure)
# ══════════════════════════════════════════════════════════
fig2, ax = plt.subplots(figsize=(13, 10))
cmap23 = plt.cm.tab20

for j, (cls, idx, auc_val) in enumerate(sub_aucs_sorted):
    fpr_i, tpr_i, _ = roc_curve(y_sub_test[:,idx], pred_sub[:,idx])
    ax.plot(fpr_i, tpr_i, color=cmap23(j / max(len(sub_aucs_sorted)-1, 1)),
            lw=1.6, label=f'{cls} ({auc_val:.3f})')

ax.plot(all_fpr_sub, mean_tpr_sub, 'k--', lw=3,
        label=f'Macro Avg (AUC={macro_auc_sub:.4f})')
ax.plot([0,1],[0,1],'k:', lw=1.2, alpha=0.4, label='Random')
ax.set_xlim([0,1]); ax.set_ylim([0,1.02])
ax.set_xlabel('False Positive Rate', fontsize=13)
ax.set_ylabel('True Positive Rate', fontsize=13)
ax.set_title('ROC Curve — Subclass Head (All Valid Classes)\nHMT-LACNN on PTB-XL Test Set',
             fontsize=14, fontweight='bold')
ax.legend(loc='lower right', fontsize=8, ncol=2,
          framealpha=0.9, edgecolor='gray')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(RESULTS_PATH + "roc_curves_subclass_all.png", dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Full subclass ROC saved")

# ══════════════════════════════════════════════════════════
# FIGURE 3: Baseline comparison ROC (all 4 models)
# ══════════════════════════════════════════════════════════
print("\nLoading baseline models for ROC comparison...")

baseline_models = {
    "SimpleCNN": BASELINE_PATH + "SimpleCNN.h5",
    "ResNet1D":  BASELINE_PATH + "ResNet1D.h5",
    "BiLSTM":    BASELINE_PATH + "BiLSTM.h5",
}
baseline_colors = {
    "HMT-LACNN (Ours)": "#E91E63",
    "ResNet1D":          "#2196F3",
    "SimpleCNN":         "#FF9800",
    "BiLSTM":            "#4CAF50",
}

fig3, ax = plt.subplots(figsize=(8, 7))

# Our model first
fpr_ours, tpr_ours, _ = roc_curve(
    y_super_test.ravel(), pred_super.ravel()
)
auc_ours = roc_auc_score(y_super_test, pred_super, average='macro')
ax.plot(fpr_ours, tpr_ours, color='#E91E63', lw=3,
        label=f'HMT-LACNN Ours (AUC={auc_ours:.4f})', zorder=5)
ax.fill_between(fpr_ours, tpr_ours, alpha=0.06, color='#E91E63')

# Baselines
for name, path in baseline_models.items():
    try:
        tf.keras.backend.clear_session()
        bm = tf.keras.models.load_model(path)
        pred_b = bm.predict(X_test_1d, batch_size=32, verbose=0)
        fpr_b, tpr_b, _ = roc_curve(y_super_test.ravel(), pred_b.ravel())
        auc_b = roc_auc_score(y_super_test, pred_b, average='macro')
        ax.plot(fpr_b, tpr_b, color=baseline_colors[name], lw=2,
                linestyle='--', label=f'{name} (AUC={auc_b:.4f})')
        print(f"  {name}: AUC={auc_b:.4f}")
    except Exception as e:
        print(f"  ⚠️  Could not load {name}: {e}")

ax.plot([0,1],[0,1],'k:', lw=1.2, alpha=0.5, label='Random Classifier')
ax.set_xlim([0,1]); ax.set_ylim([0,1.02])
ax.set_xlabel('False Positive Rate', fontsize=13)
ax.set_ylabel('True Positive Rate', fontsize=13)
ax.set_title('Baseline Comparison — ROC Curves\nPTB-XL Superclass (Micro-Average)',
             fontsize=13, fontweight='bold')
ax.legend(loc='lower right', fontsize=10.5)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(RESULTS_PATH + "roc_curves_baseline_comparison.png", dpi=150, bbox_inches='tight')
plt.show()
print(f"✅ Baseline comparison ROC saved")

print(f"\n✅ All 3 ROC figures saved to {RESULTS_PATH}")
print("  roc_curves_main.png               — 3-panel publication figure")
print("  roc_curves_subclass_all.png        — All 23 subclasses")
print("  roc_curves_baseline_comparison.png — Model comparison")
```
